Section 3: Data Quality Checks
Task 3.1 Ensure data consistency across state and district levels

```python
# 3.1.1 For each state, calculate the total number of transactions, total transaction amount, and total registered users
#by summing up the values from the district level data.
district_data_df = district_transaction_users_df.groupby(['State', 'Year', 'Quarter']).agg({
    'Transactions':'sum',
    'Amount (INR)':'sum',
    'Registered Users':'sum',
}).reset_index()

print("Total Transactions, Tranasaction Amount and Registered users for each state from district level data.")
print(district_data_df)
     
```

```python
# 3.1.2 Compare the results with the corressponding values at the state level to ensure they match.
state_data_df = state_transaction_users_df.loc[:, ['State', 'Year', 'Quarter', 'Transactions', 'Amount (INR)', 'Registered Users']]

merged_df = pd.merge(district_data_df, state_data_df, on=['State','Year', 'Quarter'], suffixes=[" District", " State"])

#Calculate discrepancies
merged_df['Transactions Diff'] = merged_df['Transactions District'] - merged_df['Transactions State']
merged_df['Amount (INR) Diff'] = merged_df['Amount (INR) District'] - merged_df['Amount (INR) State']
merged_df['Registered Users Diff'] = merged_df['Registered Users District'] - merged_df['Registered Users State']

discrepancies = merged_df[(merged_df['Transactions Diff'] != 0) | (merged_df['Amount (INR) Diff'] != 0) | (merged_df['Registered Users Diff'] != 0)]
     
```

```python
# 3.1.3 Display any discrepancies found between the district and state level data.
print("Discrepancies in State and District level data")
print(discrepancies)
```

Section 4: Data Merging and Advanced Analysis
Task 4.1 Ratio of users to population by state

```python
# 4.1.1 Merge state transactions and district demographics data to calculate the ration of registered users to the population for each state.

#Calculate population by state for district demographics data
population = district_demographics_df.groupby('State')['Population'].sum().reset_index()
users = state_transaction_users_df.groupby('State')['Registered Users'].sum().reset_index()

users_population_merged = pd.merge(users, population, on='State')

users_population_merged['Users To Population Ratio'] = users_population_merged['Registered Users'] / users_population_merged['Population']
     
```

Section 6: Insights and Conclusion

6.1 Trends and Patterns in Haryana’s Transaction Data
The digital transaction ecosystem in Haryana has witnessed substantial transformation over recent years, with noticeable structural shifts in payment behaviour. One of the most prominent trends is the rapid increase in the total transaction amount compared to the total number of transactions. This indicates not only higher adoption of digital payments but also growth in average ticket size. People are increasingly confident in making larger value payments electronically, suggesting maturation of user trust, ecosystem reliability, and platform penetration.
A critical inflection point appears post-2020. The onset of the COVID-19 pandemic acted as a catalyst for digital payment adoption. With lockdowns, mobility restrictions, and physical distancing, cash transactions were limited. Digital platforms emerged as essential alternatives for household payments, fund transfers, and essential purchases. Consequently, Haryana experienced an acceleration in both transaction frequency and value, aligning with nationwide digital adoption trends supported by government initiatives and fintech innovations.
Another significant pattern relates to transaction type contributions. In Haryana, Peer-to-Peer (P2P) payments dominate the landscape, accounting for approximately 41.5% of total transaction share in Q2 2020. This reflects the cultural and economic behaviour where personal payments, remittances, and fund transfers between individuals form a substantial proportion of digital usage. The next major contributors include Recharge and Bill Payments (30.5%) and Merchant Payments (27.6%), demonstrating that utility payments and retail transactions are increasingly digital. While merchant payments constitute a smaller proportion compared to P2P transfers, their steady growth reflects gradual onboarding of small businesses, shopkeepers, and service providers into digital ecosystems.
Overall, the trend trajectory shows a progressive shift from digital payments as an alternative option to a mainstream necessity across Haryana.

6.2 Demographic Correlation with Transaction Data
Demographic factors such as population size and density are often assumed to strongly influence digital payment activity. However, correlation analysis between Haryana’s transaction volume and population density reveals only a weak relationship (0.3075). This means that higher density areas do not automatically produce proportionally higher digital transaction volumes.
Instead, even moderately populated districts demonstrate significant digital engagement, implying that other drivers, such as fintech infrastructure, literacy levels, telecom availability, targeted government initiatives, socio-economic behaviour, and business digitization efforts influence adoption.
This weak correlation suggests that digital payment growth is not confined to urban hubs or densely populated regions. Digital financial services are reaching deeper into semi-urban and rural Haryana due to improving smartphone penetration, affordable internet, and proactive payment app outreach programs. Thus, ecosystem development is far more nuanced than population numbers alone.

6.3 Summary of Findings and Insights
The findings illustrate a fast-evolving digital landscape in Haryana, marked by rising transaction frequency and accelerated value movement. The digital economy has expanded rapidly post-2020, with total transaction volume and value surging, demonstrating increasing confidence in digital financial channels. While growth patterns emerged earlier, the pandemic amplified the need for seamless and contactless payment options, permanently altering consumer habits.
Despite this growth, population density does not strongly predict transaction volumes, indicating that digital payments are not confined to metropolitan regions. This insight highlights opportunities for policymakers, telecom operators, and fintech companies to enhance awareness, targeted interventions, and digital infrastructure in both high-density and low-density areas.
In terms of transaction categories, Peer-to-Peer payments dominate at 41.5%, reflecting a major share of value passing between individuals. Recharge/Bill payments (30.5%) and Merchant transactions (27.6%) form the next significant segments, showing that recurring household utilities and commercial digital payments are being widely adopted. These shares reflect the growing trust in bill automation, in-app payments, and small merchant QR-based transactions.
At a national comparison level, states like Telangana, Karnataka, Andhra Pradesh, Rajasthan, and Delhi emerge as top performers in average transaction value. This suggests regional variation in transaction maturity and provides Haryana benchmarks for scaling digital adoption. Another behavioural insight indicates brand dependency, with Xiaomi and Samsung devices being most widely used for making payments in Haryana, reflecting their penetration among users in different income brackets.

Recommendations
Based on the insights, several recommendations can support the long-term strengthening of Haryana’s digital payment ecosystem:
Expand Awareness Programs: Since digital adoption is strong even in less dense regions, awareness campaigns should target both rural and urban audiences. Tailored messaging can improve financial literacy and comfort with cashless transactions.
Focus on Securing P2P Channels: As P2P payments form the largest share, ensuring seamless transaction flow, fraud protection, and user confidence is critical. Enhancing authentication systems and dispute resolution mechanisms can increase trust.
Merchant and Utility Partnerships: Encouraging billers and MSMEs to adopt QR payment systems, offering cashback or loyalty incentives, can expand merchant transaction share. Supporting retailers with onboarding training and incentives can accelerate shifts towards digital cash registers.
Benchmark Against High-Performing States: Haryana should examine best practices from leading states in terms of digital incentives, infrastructure deployment, fintech collaborations, and government policies to increase average transaction size.
Deep-Dive Research Beyond Population Metrics: Further studies should incorporate factors such as income distribution, education, gender, age, occupation, mobile usage, and network coverage to build a more complete predictive model of digital adoption drivers.
Device-Linked Outreach: Since Samsung and Xiaomi dominate user devices, targeted app-based promotions, pre-installed wallet partnerships, and localized support for these brands can accelerate payment ecosystem reach.

Conclusion
Haryana’s digital payment ecosystem is expanding rapidly, driven by rising consumer comfort, external catalysts like COVID-19, and increasing fintech penetration. While population density has limited predictive power over transaction behaviour, diverse regions show strong participation, offering rich opportunities for accelerating state-wide financial inclusion. With strategic interventions, Haryana can strengthen existing growth momentum and mirror adoption patterns seen in digitally advanced states, ensuring an inclusive, secure, and scalable digital payments future.
